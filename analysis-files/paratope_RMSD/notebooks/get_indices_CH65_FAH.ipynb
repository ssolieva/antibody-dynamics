```python
BASE_pdb = '/mnt/pure/bowmanlab/WG-shahlo/projects/Ab_dynamics/FAH-analysis-20260212/clustering/radius_15_variable_subsample50/data/apo/representative_clusters/top_cluster_centers/'
```

```python
# Naive

sys_name = 'CH65_naive_VC'

partial_seq_list = [
    'QVQLVQSGAEVKKPGASVKVSCKAS',
    'MHWVRQAPGQGLEWMGW',
    'NYAQKFQGWVTMTRDTSISTAYMELSRLRSDDTAVYYC',
    'WGQGTTVTVSS',
    'SVLTQPPSVSVAPGQTARITCGGN',
    'VHWYQQKPGQAPVLVVY',
    'DRPSGIPERFSGSNSGNTATLTISRVEAGDEADYYC',
    'FGGGTKLTVL']

structure_path = f'{BASE_pdb}{sys_name}-topcenter.pdb'
sequence, ca_indices =FR_indices_toolbox.get_sequence(structure_path)

FR_indices = []
for partial_seq in partial_seq_list:
    partial_seq_ca_indices = FR_indices_toolbox.find_indices(sequence, ca_indices, partial_seq)
    FR_indices_toolbox.partial_seq_check(structure_path, partial_seq_ca_indices, partial_seq)
    FR_indices.extend(partial_seq_ca_indices)

print(f'Number of atoms: {len(FR_indices)}')
np.save(f'../data/{sys_name}-FR-indices-FAH.npy', FR_indices)
```
