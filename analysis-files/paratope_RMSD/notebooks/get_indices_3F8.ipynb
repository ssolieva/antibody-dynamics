```python
# Reference 

partial_seq_list = [
'QVQLKESGPGLVAPSQSLSITCTVS',
'VHWVRQPPGKGLEWLGV',
'NYNSAFMSRLSISKDNSKSQVFLKMNSLQIDDTAMYYC',
'WGQGTSVTVSS',
'SIVMTQTPKFLLVSAGDRVTITCKAS',
'VTWYQQKAGQSPKLLIY',
'NRYSGVPDRFTGSGYGTAFTFTISTVQAEDLAVYFC',
'FGGGTKL']

structure_path = '/mnt/pure/bowmanlab/WG-shahlo/projects/Ab_dynamics/analysis/paratope_RMSD/ref_structures/3vfg_mature_3f8.pdb'
sequence, ca_indices =FR_indices_toolbox.get_sequence(structure_path)

FR_indices = []
for partial_seq in partial_seq_list:
    partial_seq_ca_indices = FR_indices_toolbox.find_indices(sequence, ca_indices, partial_seq)
    FR_indices_toolbox.partial_seq_check(structure_path, partial_seq_ca_indices, partial_seq)
    FR_indices.extend(partial_seq_ca_indices)

print(f'Number of atoms: {len(FR_indices)}')
np.save(f'../data/3F8-ref-FR-indices.npy', FR_indices)
```
